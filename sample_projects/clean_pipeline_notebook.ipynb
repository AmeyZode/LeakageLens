# Production Customer Churn Classifier (Clean Pipeline)
This notebook demonstrates a clean ML pipeline with zero data leakage, explicit reproducibility, and strict validation isolation.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Initialize random seeds for reproducibility
np.random.seed(42)

# 2. Load dataset using relative path
dataset = pd.read_csv('sample_projects/clean_customer_data.csv')

# 3. Extract features and labels
feature_columns = ['age', 'account_balance', 'tenure', 'transaction_count', 'satisfaction_score']
features = dataset[feature_columns]
labels = dataset['churned']
```

```python
# 4. Split dataset before applying any transformations
features_train, features_test, labels_train, labels_test = train_test_split(
    features,
    labels,
    test_size=0.2,
    random_state=42
)

# 5. Fit scaler on training partition only, transform both partitions
scaler = StandardScaler()
scaled_train = scaler.fit_transform(features_train)
scaled_test = scaler.transform(features_test)
```

```python
# 6. Train model with explicit random_state
classifier = RandomForestClassifier(n_estimators=50, random_state=42)
classifier.fit(scaled_train, labels_train)

# 7. Evaluate exclusively on out-of-sample test partition
predictions = classifier.predict(scaled_test)
accuracy = accuracy_score(labels_test, predictions)
print(f'Test Accuracy: {accuracy:.4f}')
```
